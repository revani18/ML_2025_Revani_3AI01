# Import Library

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

# Load Data

```python
df_3 = pd.read_csv(path + "/Data/dataset_satelit.csv")
df_3.head()
```

Data tentang Satelit, memprediksi nitrogen berdasarkan fitur yang ada di datasetnya

Interpretasi Kolom

```python
#ubah kolom mg menjadi float
df_3['Mg'] = pd.to_numeric(df_3['Mg'], errors='coerce')

#hilangkan baris yang memiliki missing value
df_3 = df_3.dropna()
```

# Feature Selection

```python
# Memilih fitur dengan beberapa kolom
# X_3 = df_3[['b2', 'b3', 'b4', 'b8', 'b11', 'Sigma_VV', 'Sigma_VH']]
# y_3 = df_3['N']
```

```python
# Memilih fitur dengan semua kolom
X_3 = df_3[['P', 'K', 'Ca', 'Mg', 'Fe', 'Mn', 'Cu', 'Zn', 'B',
            'b12', 'b11', 'b9', 'b8a', 'b8', 'b7', 'b6', 'b5',
            'b4', 'b3', 'b2', 'b1', 'Sigma_VV', 'Sigma_VH', 'plia', 'lia', 'iafe',
            'gamma0_vv', 'gamma0_vh', 'beta0_vv', 'beta0_vh']]
y_3 = df_3['N']
```

# Splitting Data

Pembagian data yang menggunakan semua kolom

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_3, y_3, test_size=0.2, random_state=42
)
```

```python
model = LinearRegression()

#training data
model.fit(X_train, y_train)
```

Multiple Linear Regression, karena fitur lebih dari 1

```python
#testing
y_pred = model.predict(X_test)

#evaluasi
r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("R2 Score:", r2)
print("RMSE:", mse)
```

Jika RMSE makin beda maka modelnya jelek

```python
coeff = pd.DataFrame({
    'Fitur': X_3.columns,
    'Koefisien': model.coef_
})

print(coeff)
```

```python
import statsmodels.api as sm

X_sm = sm.add_constant(X_3)
model_ols = sm.OLS(y_3, X_sm).fit()
print(model_ols.summary())
```
